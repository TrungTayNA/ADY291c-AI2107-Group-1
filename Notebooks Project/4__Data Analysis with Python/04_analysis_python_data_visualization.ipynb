# Phân tích và trực quan hóa trải nghiệm của khách hàng Olist
Notebook phân tích Python và trực quan hóa thành một luồng cho 8 câu hỏi. Mỗi phần lần lượt trình bảngm kết quả phân tích, nhận xét và lựa chọn bảng dùng để vẽ, một cell trực quan hóa, rồi nhận xét biểu đồ. Phần kết luận tổng nằm cuối notebook.

### Mục tiêu phân tích
- **Câu 1:** Đánh giá phân bố điểm đánh giá và mức độ mất cân bằng giữa các nhóm điểm.
- **Câu 2:** So sánh mức hài lòng giữa đơn giao trễ và đơn giao đúng hạn.
- **Câu 3:** Xác định mức độ giao trễ liên quan như thế nào đến điểm đánh giá.
- **Câu 4:** Khảo sát mối liên hệ giữa phí vận chuyển, tỷ lệ phí trên giá hàng và điểm đánh giá; xem xét tương quan với các đặc trung vận hành.
- **Câu 5:** Đánh giá mối liên hệ giữa độ phức tạp đơn hàng và điểm đánh giá.
- **Câu 6:** So sánh điểm đánh giá theo phương thức thanh toán và số kỳ trả góp.
- **Câu 7:** Theo dõi xu hướng điểm đánh giá theo tháng đặt hàng.
- **Câu 8:** So sánh mức hài lòng giữa các bang khách hàng

Dữ liệu lấy từ view `v_review_order_features`, với mỗi dòng cho mỗi đơn hàng có review. Các kết quả là mô tả mối liên hệ, không chứng minh quan hệ nhân quả. Điểm thấp được định nghĩa là 1-2 sao; khi đánh giá mô hình, cần phân biệt có sẵn lúc đặt hàng với biến chỉ biết sau giao hàng để tránh rò rỉ dữ liệu.

```python
# Import thư viện dùng để đọc dữ liệu, phân tích và trực quan hóa.
from pathlib import Path 
import sqlite3

import matplotlib.pyplot as plt
plt.rcParams["font.family"] = "Arial"
plt.rcParams["axes.unicode_minus"] = False
import numpy as np 
import pandas as pd 
import seaborn as sns
from scipy import stats

# Tìm thư mục gốc để notebook chạy được dù được mở từ thư mục con.
def find_project_root(start_dir: Path) -> Path:
    for candidate in (start_dir, *start_dir.parents):
        if (
            (candidate / "data" / "olist.db").is_file()
            and (candidate / "sql" / "03_queries.sql").is_file()
            and (candidate / "Notebooks").is_dir()
        ):
            return candidate
    raise FileNotFoundError(
        f"Không tìm thấy thư mục gốc dự án {start_dir}. "
        "Hãy chạy scripts/database/build_db.py trước để tạo data/olist.db"
    )

# Khai báo đường dẫn database và thư mục lưu biểu đồ.
BASE_DIR = find_project_root(Path.cwd().resolve())
DB_PATH = BASE_DIR / "data" / "olist.db"
FIGURES_DIR = BASE_DIR / "report" / "figures"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

# Lưu biểu đồ hiện tại vào thư mục báo cáo với độ phân giải phù hợp.
def save_figure(filename: str) -> None:
    plt.gcf().savefig(FIGURES_DIR / filename, dpi=150, bbox_inches= "tight")
    
# Định dạng p-value gọn hơn khi in kết quả kiểm định thống kê.
def format_p_value(p_value: float) -> str:
    return "<0.001" if p_value < 0.001 else f"{p_value:.3g}"

# Kiểm tra view cần dùng đã tồn tại rồi mới đọc dữ liệu vào DataFrame.
with sqlite3.connect(DB_PATH) as conn:
    available_views = {
        row[0]
        for row in conn.execute(
            "SELECT name FROM sqlite_master WHERE type = 'view'"
        )
    }
    if "v_review_order_features" not in available_views:
        raise RuntimeError(
            "Database chưa có view v_review_order_features. "
            "Hãy chạy scripts/database/build_db.py để dựng lại database."
        )
    review_orders = pd.read_sql_query(
        "SELECT * FROM v_review_order_features ORDER BY purchase_year_month, order_id",
        conn,
    )
    
# Chuẩn hóa kiểu dữ liệu và tạo cờ cho nhóm review thấp/tích cực.
review_orders["is_late"] = review_orders["is_late"].astype(bool)
review_orders["review_score"] = review_orders["review_score"].astype(int)
review_orders["low_score"] = review_orders["review_score"] <= 2
review_orders["positive_review"] = review_orders["review_score"] >= 4
review_orders["purchase_year_month"] = pd.to_datetime(
    review_orders["purchase_year_month"], format="%Y-%m"
)
# Mức ý nghĩa dùng cho các kiểm định thống kê ở những phần sau.
ALPHA = 0.05
print(f"Database: {DB_PATH}")
print(f"Đơn hàng có review trong view SQL: {len(review_orders):,}")
print("Thang điểm review", sorted(review_orders["review_score"].unique()))
```

# Câu 1 - Phân bố và mất cân bằng điểm đánh giá
Đếm số đơn và tỷ lệ của từng mức 1-5 sao để kiểm tra phân bố nhãn và xác định như cầu theo dõi nhóm review thấp

```python
# Đếm số đơn theo từng mức review; reindex để luôn hiển thị đủ 1-5 sao.
q01 = (
    review_orders.groupby("review_score").size()
    .reindex(range(1, 6), fill_value=0)
    .rename("order_count")
    .to_frame()
)
# Tính tỷ trọng của mỗi mức điểm trên tổng số đơn có review.
q01["percentage"] = q01["order_count"] / q01["order_count"].sum() * 100
q01["percentage"] = q01["percentage"].round(2)
# Hiển thị bảng phân bố điểm.
display(q01)

# Tóm tắt điểm trung bình và quy mô nhóm review thấp (1-2 sao).
average_review_score = review_orders["review_score"].mean()
low_score_count = q01.loc[1:2, "order_count"].sum()
low_score_percentage = low_score_count / q01["order_count"].sum() * 100

print(f"Điểm đánh giá trung bình: {average_review_score:.2f}/5")
print(
    f"Tỷ lệ đánh giá 1-2 sao: {low_score_percentage:.2f}% "
    f"({low_score_count:,}/{q01['order_count'].sum():,} đơn hàng)"
)
```

**Nhận xét phân tích và chọn bảng vẽ**: Điểm trung bình chung là 4,16/5; 5 sao chiếm 59,25% (56,480 đơn), còn nhóm 1-2 sao chiếm 12,77%. Dùng bảng `q01`, cụ thể là số đơn và tỷ lệ theo từng mức sao, để vẽ một biểu đồ cột thể hiện đồng thời tỷ trọng từng lớp.

```python
# Tạo biểu đồ cột với số đơn theo từng mức điểm.
fig, ax = plt.subplots(figsize = (8, 4.5))
bars = ax.bar(q01.index.astype(str), q01["order_count"], color= "#2878B5", label = "Số đơn có review")
# Ghi tỷ lệ phần trăm lên đầu cột để đọc được cả số lượng và tỷ trọng.
ax.bar_label(bars, labels=[f"{value:.1f}%" for value in q01["percentage"]], padding=3)
ax.set(title = "Câu 1 - Phân bố điểm đánh giá", xlabel = "Điểm review (sao)", ylabel = "Số đơn có review")
ax.legend(loc= "upper right")
ax.margins(y= 0.15)
fig.tight_layout()
# Lưu biểu đồ để dùng lại trong báo cáo, sau đó hiển thị và giải phóng figure.
save_figure("04_01_review_score_distribution.png")
plt.show()
plt.close(fig)
```

**Nhận xét biểu đồ:** Phân bố nghiêng về điểm cao: 5 sao chiếm 59,25%, nhưng nhóm 1-2 sao vẫn chiếm 12,77% số đơn có review. Vì vậy, điểm trung bình chung 4,16/5 không phản ánh đầy đủ quy mô khách hàng không hài lòng. Olist cần theo dõi riêng nhóm điểm thấp và dùng thước đo phù hợp với nhãn mất cân bằng khi xây dựng cảnh báo sớm.

## Câu 2 - Giao trễ và giao đúng hạn liên quan đến điểm đánh giá thế nào?

So sánh điểm trung bình, tỷ lệ đánh giá thấp (<= 2 sao), đánh giá tích cực (>= 4 sao) và cỡ mẫu giữa nhóm giao trễ với nhóm giao đúng hạn.

```python
# Chia các đơn hàng theo is_late: False = đúng hạn/sớm, True = giao trễ.
q02 = review_orders.groupby("is_late").agg(
    # size đếm số đơn hàng trong mỗi nhóm.
    order_count=("order_id", "size"),
    # mean tính điểm review trung bình của mỗi nhóm.
    avg_review_score=("review_score", "mean"),
    # Hai cột positive_review và low_score là True/False; mean của chúng chính là tỷ lệ.
    positive_pct =("positive_review", "mean"),
    low_score_pct =("low_score","mean"),
    # Đổi nhãn nhóm False/True thành tên dễ đọc hơn.
).rename(index={False: "Đúng hạn hoặc sớm", True: "Giao Trễ"})
# Chuyển tỷ lệ dạng số thập phân (ví dụ 0.8271) sang phần trăm (82.71).
q02[["positive_pct", "low_score_pct"]] *= 100
# Làm tròn các giá trị số đến 2 chữ số thập phân.
q02 = q02.round(2)
# Hiển thị bảng kết quả trong notebook.
display(q02)
```

**Nhận xét phân tích và chọn bảng vẽ:** Nhóm giao trễ có điểm trung bình 2,27, thấp hơn nhóm đúng hạn hoặc sớm (4,29); tỷ lệ review thấp lần lượt là 62,37% và 9,23%. Chọn bảng `q02` vì bảng có cả điểm trung bình, tỷ lệ review tích cực/thấp và số đơn; vẽ 2 biểu đồ 1 biểu đồ cột đơn hiện thị tỉ lệ đánh giá trung bình và 1 biểu đồ cột nhóm tỷ lệ đánh giá theo nhóm

```python
# Tạo hai biểu đồ cạnh nhau để so sánh điểm trung bình và tỷ lệ review.
fig, axes = plt.subplots(1,2,figsize=(11, 4.5))
# Panel trái: điểm review trung bình theo tình trạng giao hàng.
q02["avg_review_score"].plot(kind="bar", ax=axes[0], color=["#4C9F70", "#D9534F"], label= "Điểm review trung bình")
axes[0].set(title="Điểm đánh giá trung bình", ylabel="Điểm sao", xlabel="Tình trạng giao hàng", ylim=(0, 5))
axes[0].tick_params(axis="x", rotation=15)
axes[0].legend(loc="upper right")
# Panel phải: đặt tỷ lệ review tích cực và review thấp cạnh nhau.
q02[["positive_pct", "low_score_pct"]].plot(kind="bar", ax=axes[1], color=["#4C9F70", "#D9534F"])
axes[1].set(title="Tỷ lệ review theo nhóm", ylabel= "Tỷ lệ review (%)", xlabel= "Tình trạng giao hàng")
axes[1].tick_params(axis="x", rotation=15)
axes[1].legend(["Tích cực (>= 4)", "Thấp (<= 2)"])
fig.suptitle("Câu 2 - Tình trạng giao hàng và đánh giá điểm")
fig.tight_layout()
# Lưu và hiển thị biểu đồ so sánh của Câu 2.
save_figure("04_02_late_vs_review_score.png")
plt.show()
plt.close(fig)
```

**Nhận xét:** Đơn giao trễ có điểm trung bình 2,27/5 và tỷ lệ review thấp 62,37%, so với 4,29/5 và 9,23% ở nhóm giao đúng hạn hoặc sớm. Chênh lệch lớn cho thấy tình trạng giao hàng là tín hiệu vận hành đáng ưu tiên theo dõi, dù đây chỉ là mối liên hệ chứ không chứng minh nguyên nhân. Phát hiện sớm đơn có nguy cơ trải nghiệm xấu để Olist chủ động phối hợp logistics và chăm sóc khách hàng.

## Câu 3 - Mức độ giao trễ và review score

Chia `delay_days` thành bốn nhóm: sớm/đúng hạn, trễ 1-3 ngày, trễ 4-7 ngày và trễ trên 7 ngày; so sánh điểm trung bình, tỷ lệ đánh giá thấp/tích cực và số đơn.

```python
# Khai báo tên bốn nhóm độ trễ theo đúng thứ tự cần trình bày.
delay_labels = ["Sớm hoặc đúng hạn", "Trễ 1-3 ngày", "Trễ 4-7 ngày", "Trễ trên 7 ngày"]
# Chia delay_days theo các ngưỡng 0, 3 và 7 ngày; gán nhãn tương ứng.
review_orders["delay_group"] = pd.cut(
    review_orders["delay_days"], bins= [-np.inf, 0, 3, 7 , np.inf], labels=delay_labels
)
# Tính số đơn, điểm trung bình và tỷ lệ review thấp/tích cực trong mỗi nhóm.
q03 = review_orders.groupby("delay_group", observed=False).agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
    positive_pct = ("positive_review", "mean"),
)
# Mean của cờ Boolean là tỷ lệ; đổi sang phần trăm và làm tròn hai chữ số.
q03[["low_score_pct", "positive_pct"]]*= 100
q03 = q03.round(2)
# Hiển thị bảng kết quả theo nhóm độ trễ.
display(q03)
```

**Nhận xét phân tích và chọn bảng vẽ:** Điểm trung bình giảm theo mức trễ: 3,29 ở nhóm trễ 1-3 ngày, 2,11 ở nhóm trễ 4-7 ngày và 1,69 ở nhóm trễ trên 7 ngày; tỷ lệ review thấp đạt 79,31% ở nhóm trễ trên 7 ngày. Chọn bảng `q03` và vẽ điểm trung bình theo nhóm trễ.

```python
# Vẽ điểm trung bình theo từng nhóm độ trễ đã tính ở bảng q03.
fig, ax = plt.subplots(figsize=(9, 4.5))
bars = ax.bar(q03.index.astype(str), q03["avg_review_score"], color="#2878B5", label= "Điểm đánh giá trung bình")
# Ghi điểm trung bình và số đơn lên từng cột để người đọc xét cả cỡ mẫu.
ax.bar_label(bars, labels=[f"{score:.2f} (n={count:,})" for score, count in zip(q03["avg_review_score"], q03["order_count"])], padding = 3, fontsize = 8)
ax.set(title="Câu 3 - Điểm đánh giá theo số ngày trễ", xlabel="Mức trễ", ylabel="Điểm đánh giá trung bình", ylim=(0, 5))
ax.tick_params(axis="x", rotation = 15)
ax.grid(axis="y", alpha=0.25)
ax.legend(loc="upper right")
fig.tight_layout()
# Lưu biểu đồ, hiển thị trong notebook rồi đóng figure.
save_figure("04_03_delay_vs_review_score.png")
plt.show()
plt.close(fig)
```

**Nhận xét:** Điểm trung bình giảm từ 3,29 ở nhóm trễ 1-3 ngày xuống 2,11 ở nhóm trễ 4-7 ngày và 1,69 ở nhóm trễ trên 7 ngày. Mức giảm theo độ trễ gợi ý rủi ro không chỉ nằm ở trạng thái trễ mà còn tăng theo số ngày chậm; các nhóm trễ có ít đơn hơn nên cần đọc cùng cỡ mẫu. Theo dõi các ngưỡng trễ giúp Olist ưu tiên can thiệp theo mức độ cấp bách thay vì xử lý mọi đơn như nhau.

### Phân tích bổ sung — phân bố độ trễ theo nhóm review

```python
# Tạo nhóm rủi ro review để đối chiếu phân bố số ngày giao so với hạn.
delay_plot_orders = review_orders.assign(
    review_risk_group=np.where(
        review_orders["review_score"].le(2), "Review thấp (1-2)", "Review không thấp (3-5)"
    )
)
# Cố định thứ tự nhóm để bảng và biểu đồ luôn hiển thị nhất quán.
risk_order = ["Review thấp (1-2)", "Review không thấp (3-5)"]
# Tóm tắt số đơn, trung vị và hai tứ phân vị của độ trễ trong từng nhóm.
delay_by_risk = (
    delay_plot_orders.groupby("review_risk_group", observed=True)
    .agg(
        order_count=("order_id", "size"),
        median_delay_days=("delay_days", "median"),
        q1_delay_days=("delay_days", lambda values: values.quantile(0.25)),
        q3_delay_days=("delay_days", lambda values: values.quantile(0.75)),
    )
    .reindex(risk_order)
)
# Hiển thị các thống kê đã làm tròn để so sánh mức độ và độ phân tán.
print("Số ngày chênh so với hạn giao dự kiến theo nhóm review")
display(delay_by_risk.round(2))
```

```python
# Dùng boxplot để so sánh phân bố độ trễ giữa hai nhóm review.
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(
    data=delay_plot_orders,
    x="review_risk_group",
    y="delay_days",
    hue="review_risk_group",
    order=risk_order,
    hue_order=risk_order,
    palette={"Review thấp (1-2)": "#D9534F","Review không thấp (3-5)": "#4C9F70",},
    dodge=False,
    showfliers=False,
    ax=ax,
)
# Mốc 0 biểu thị đúng hạn; giá trị âm là sớm và dương là trễ.
ax.axhline(0, color="#555555", linestyle="--", linewidth=1, label="Đúng hạn")
ax.set(
    title="Độ trễ giao hàng theo nhóm đánh giá",
    xlabel="Nhóm điểm đánh giá",
    ylabel="Số ngày so với hạn giao dự kiến (ngày; âm = sớm)",
)
# Tạo chú giải tùy chỉnh cho màu hai nhóm và đường mốc đúng hạn.
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

ax.legend(
    handles=[
        Patch(facecolor="#D9534F", label="Đánh giá thấp (1-2)"),
        Patch(facecolor="#4C9F70", label="Đánh giá không thấp (3-5)"),
        Line2D([0], [0], color="#555555", linestyle="--", label="Đúng hạn"),
    ],
    title="Nhóm đánh giá / mốc hạn",
    loc="best",
)
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
# Lưu, hiển thị và đóng biểu đồ boxplot.
save_figure("04_03_delivery_delay_by_review_risk.png")
plt.show()
plt.close(fig)
```

**Nhận xét:** Boxplot so sánh số ngày chênh lệch với hạn giao dự kiến giữa hai nhóm review; giá trị âm biểu thị giao sớm, còn dương biểu thị giao trễ. Trung vị của nhóm review thấp là -8 ngày, so với -13 ngày ở nhóm review không thấp, tức nhóm thấp nhìn chung giao ít sớm hơn; độ phân tán cũng cho biết mức độ khác biệt bên trong từng nhóm. Hai phân bố vẫn chồng lấn, nên giao trễ không đồng nghĩa chắc chắn với đánh giá thấp.

## Câu 4 - Tổng phí vận chuyển, tỷ lệ phí vận chuyển/giá trị hàng và điểm đánh giá

Chia đơn thành 4 nhóm tứ phân vị theo tổng phí vận chuyển và tỷ lệ phí vận chuyển/giá trị hàng. So sánh đánh giá giữa các nhóm và dùng Spearman để đo liên hệ đơn điệu; đơn có giá trị hàng bằng 0 không tham gia phân tích tỷ lệ phí vận chuyển/giá trị hàng.

```python
# Đặt nhãn các nhóm tứ phân vị từ mức phí thấp đến cao.
quartile_labels = ["Q1 thấp", "Q2", "Q3", "Q4 cao"]
# Chia cả tổng phí vận chuyển và tỷ lệ phí/giá thành bốn nhóm.
# Xếp hạng trước khi qcut để xử lý được các giá trị trùng lặp tại ranh giới.
for source, target in [
    ("total_freight", "freight_quartile"),
    ("freight_to_price_pct", "freight_ratio_quartile"),
]:
    review_orders[target] = pd.qcut(
        review_orders[source].rank(method="first"), q=4, labels=quartile_labels
    )
# Tổng hợp quy mô mẫu, phí trung bình, điểm và tỷ lệ review thấp theo quartile phí.
q04_freight = review_orders.groupby("freight_quartile", observed=False).agg(
    order_count=("order_id", "size"),
    avg_freight=("total_freight", "mean"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
)
# Bỏ đơn thiếu tỷ lệ phí/giá trước khi thống kê theo quartile tương ứng.
valid_ratio = review_orders.dropna(subset=["freight_to_price_pct"])
# Tổng hợp các chỉ số tương tự cho quartile tỷ lệ phí trên giá trị hàng.
q04_ratio = valid_ratio.groupby("freight_ratio_quartile", observed=False).agg(
    order_count=("order_id", "size"),
    avg_freight_ratio_pct=("freight_to_price_pct", "mean"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
)
# Chuyển tỷ lệ review thấp từ dạng thập phân sang phần trăm.
q04_freight["low_score_pct"] *= 100
q04_ratio["low_score_pct"] *= 100
# Đo tương quan Spearman giữa từng cách đo phí và điểm review.
freight_rho, freight_p = stats.spearmanr(review_orders["total_freight"], review_orders["review_score"])
ratio_rho, ratio_p = stats.spearmanr(valid_ratio["freight_to_price_pct"], valid_ratio["review_score"])
# Hiển thị các bảng so sánh cùng hệ số và p-value của kiểm định Spearman.
print("Theo tổng phí vận chuyển:")
display(q04_freight.round(2))
print("Theo tỷ lệ phí vận chuyển/giá trị hàng:")
display(q04_ratio.round(2))
print(f"Spearman tổng phí ship và review: rho={freight_rho:.3f}, p={format_p_value(freight_p)}")
print(f"Spearman tỷ lệ phí/giá và review: rho={ratio_rho:.3f}, p={format_p_value(ratio_p)}")
```

**Nhận xét phân tích và chọn bảng vẽ:** Q4 phí vận chuyển cao nhất có điểm trung bình 3,94, thấp hơn Q1 thấp nhất (4,32); theo tỷ lệ phí vận chuyển/giá trị hàng, chênh lệch nhỏ hơn (4,20 so với 4,11). Spearman lần lượt là -0,088 và -0,03, tức chiều liên hệ âm nhưng yếu. Dùng hai bảng `q04_freight` và `q04_ratio` trong một hình hai panel, cùng thang điểm, để đối chiếu hai cách đo phí.

```python
# So sánh điểm trung bình theo hai cách chia nhóm phí trên cùng thang điểm.
fig, axes = plt.subplots(1,2, figsize= (11, 4.5), sharey= True)
# Ghép từng bảng tóm tắt với panel và tiêu đề tương ứng.
for ax, summary, title in [ 
    (axes[0], q04_freight, "Theo tổng phí vận chuyển"),
    (axes[1], q04_ratio , "Theo tỷ lệ phí / giá"),
]:
    # Vẽ điểm trung bình; ghi cả cỡ mẫu lên cột để tiện đối chiếu.
    bars = ax.bar(summary.index.astype(str), summary["avg_review_score"], color = "#8172B2", label = "Điểm review trung bình")
    ax.bar_label(bars, labels=[f"{score:.2f} (n={count:,})" for score, count in zip (summary["avg_review_score"], summary["order_count"])], padding= 3, fontsize= 8)
    ax.set(title=title, xlabel="Nhóm phân vị từ thấp đến cao", ylim = (0,5))
    ax.tick_params(axis="x", rotation=15)
    ax.grid(axis="y", alpha = 0.25)
    ax.legend(loc="upper right")
# Đặt nhãn chung và hoàn thiện, lưu biểu đồ của Câu 4.
axes[0].set_ylabel("Điểm đánh giá sao trung bình")
fig.suptitle("Câu 4 - Phí vận chuyển và điểm đánh giá")
fig.tight_layout()
save_figure("04_04_freight_vs_review_score.png")
plt.show()
plt.close(fig)   
    
```

**Nhận xét** Điểm trung bình ở quartile phí vận chuyển cao nhất là 3,94, thấp hơn 4,32 ở quartile thấp nhất; chênh lệch theo tỷ lệ phí/giá còn nhỏ hơn. P-value nhỏ trên tập lớn không đồng nghĩa tác động thực tế mạnh, nhất là khi tương quan đơn biến với review yếu. Do đó, phí vận chuyển không nên được dùng riêng để giải thích trải nghiệm; cần xem cùng độ trễ, khu vực và độ phức tạp của đơn hàng.

### Ma trận tương quan giữa đánh giá và đặc trưng vận hành

Sử dụng tương quan Spearman vì điểm đánh giá là thang thứ bậc và các đại lượng tiền có phân bố lệch. Ma trận giúp so sánh đồng thời các quan hệ với review cũng như sự phụ thuộc giữa chính các biến giải thích; hệ số nhỏ không loại trừ các ngưỡng rủi ro hoặc tương tác giữa nhóm. `delay_days` và các trường giao hàng chỉ phù hợp cho phân tích sau giao hàng, không phải dự báo tại thời điểm đặt hàng.

```python
# Chọn các đặc trưng số cần so sánh với review score.
correlation_columns = [
    "review_score",
    "delay_days",
    "total_freight",
    "freight_to_price_pct",
    "item_count",
    "seller_count",
    "total_paid",
    "max_installments"
    
]
# Tính tương quan Spearman để đo mức liên hệ đơn điệu giữa các biến.
correlation_matrix = review_orders[correlation_columns].corr(method="spearman")
print("Ma trận tương quan Spearman (p)") 
display(correlation_matrix.round(2))
# Tìm biến có trị tuyệt đối tương quan lớn nhất với review score.
review_correlations = correlation_matrix["review_score"].drop("review_score")
strongest_factor = review_correlations.abs().idxmax()
print(
    "Liên hệ đơn điệu mạnh nhất với review score trong các biến đã chọn: "
    f"{strongest_factor}, p={review_correlations[strongest_factor]:.3f}."
)
```

Số mặt hàng, số người bán và phí vận chuyển cũng chỉ có tương quan đơn biến yếu với điểm. Vì thế, bảng chéo theo nhóm giao hàng và độ phức tạp bổ sung tín hiệu tương tác mà một ma trận tương quan đơn lẻ không thể hiện; kết quả định hướng theo dõi logistics nhưng không chứng minh quan hệ nhân quả.

## Câu 5 — Độ phức tạp đơn hàng và đánh giá điểm sao

Đối chiếu mức đánh giá theo số mặt hàng, số sản phẩm khác nhau và số người bán trong đơn.

```python
# Gom số mặt hàng, số sản phẩm khác nhau và số người bán thành nhóm.
review_orders["basket_group"] = np.select(
    [review_orders["item_count"].eq(1), review_orders["item_count"].between(2, 3)],
    ["1 mặt hàng", "2-3 mặt hàng"], default="4+ mặt hàng"
)
review_orders["product_group"] = np.select(
    [review_orders["distinct_products"].eq(1), review_orders["distinct_products"].between(2, 3)],
    ["1 sản phẩm", "2-3 sản phẩm"], default="4+ sản phẩm"
)
review_orders["seller_group"] = np.where(
    review_orders["seller_count"].eq(1), "1 người bán", "Nhiều người bán"
)
# Tạo bảng thống kê số đơn, điểm trung bình và tỷ lệ review thấp theo từng tiêu chí.
q05_tables = {}
for column, title in [
    ("basket_group", "Số mặt hàng"),
    ("product_group", "Số sản phẩm khác nhau"),
    ("seller_group", "Số người bán"),
]:
    summary = review_orders.groupby(column).agg(
        order_count=("order_id", "size"),
        avg_review_score=("review_score", "mean"),
        low_score_pct=("low_score", "mean"),
    )
    # Đổi tỷ lệ sang phần trăm, làm tròn và lưu để dùng tiếp khi vẽ.
    summary["low_score_pct"] *= 100
    q05_tables[title] = summary.round(2)
    print(title)
    display(q05_tables[title])
```

**Nhận xét phân tích và chọn bảng vẽ:** Đơn một mặt hàng có điểm trung bình **4,21**, so với **3,36** ở đơn từ 4 mặt hàng; đơn nhiều người bán có điểm **2,86** và tỷ lệ review thấp **47,36%**, so với **4,17** và **12,31%** ở đơn một người bán. Chọn ba bảng trong `q05_tables` để đặt trong một hình ba panel.

```python
# Vẽ ba góc nhìn về độ phức tạp của đơn hàng cạnh nhau.
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
for ax, (title, summary) in zip(axes, q05_tables.items()):
    # Hiển thị điểm trung bình và số đơn của từng nhóm.
    bars = ax.bar(summary.index.astype(str), summary["avg_review_score"], color="#2878B5", label="Điểm đánh giá trung bình")
    ax.bar_label(bars, labels=[f"{score:.2f} (n={count:,})" for score, count in zip(summary["avg_review_score"], summary["order_count"])], padding=2, fontsize=7)
    ax.set(title=title, xlabel="Nhóm đơn hàng", ylim=(0, 5))
    ax.tick_params(axis="x", rotation=25)
    ax.grid(axis="y", alpha=0.25)
    ax.legend(loc="upper right")
# Dùng chung tiêu đề/trục dọc rồi lưu hình Câu 5.
axes[0].set_ylabel("Điểm đánh giá trung bình (1-5)")
fig.suptitle("Câu 5 — Độ phức tạp đơn hàng và điểm đánh giá")
fig.tight_layout()
save_figure("04_05_order_complexity_vs_review_score.png")
plt.show()
plt.close(fig)
```

**Nhận xét:** Điểm trung bình giảm từ 4,21 ở đơn một mặt hàng xuống 3,36 ở nhóm từ bốn mặt hàng; đơn có nhiều người bán có điểm trung bình 2,86 và tỷ lệ review thấp cao hơn. Điều này gợi ý rủi ro phối hợp có thể tồn tại bên cạnh bản thân thời gian giao hàng, nhưng các nhóm hiếm có cỡ mẫu nhỏ và kết quả chỉ mang tính mô tả. Olist nên theo dõi riêng đơn nhiều mặt hàng/người bán để kiểm tra khả năng phát sinh trải nghiệm phân mảnh và điều phối giao nhận phức tạp.

### Bảng chéo và nhóm đa chỉ mục - kết hợp độ trễ với số người bán

Bảng chéo dưới đây đối chiếu đồng thời mức độ giao trễ, số người bán và nhóm đánh giá; bảng chuẩn hóa cho biết tỷ lệ đánh giá thấp trong từng tổ hợp. Nhóm đa chỉ mục bổ sung quy mô mẫu, điểm trung bình và tỷ lệ đánh giá thấp để phân biệt tín hiệu có nhiều quan sát với những nhóm hiếm. Đây là phân tích mô tả có điều kiện, không chứng minh quan hệ nhân quả.

```python
# Tạo bản sao để thêm các nhóm phân tích mà không đổi dữ liệu gốc.
analysis_orders = review_orders.copy()
# Phân nhóm độ trễ và số người bán cho bảng chéo.
analysis_orders["low_score"] = analysis_orders["review_score"].le(2)
analysis_orders["delivery_band"] = pd.cut(
    analysis_orders["delay_days"],
    bins=[-np.inf, 0, 3, 7, np.inf],
    labels=["Đúng hạn/sớm", "Trễ 1–3 ngày", "Trễ 4–7 ngày", "Trễ trên 7 ngày"],
    include_lowest=True,
)
analysis_orders["seller_group"] = np.where(
    analysis_orders["seller_count"].gt(1), "Từ 2 người bán", "1 người bán"
)

# Đếm số đơn và tính tỷ lệ review thấp/không thấp theo từng tổ hợp nhóm.
cross_index = ["delivery_band", "seller_group"]
crosstab_counts = pd.crosstab(
    [analysis_orders[column] for column in cross_index],
    analysis_orders["low_score"],
    dropna=False,
).rename(columns={False: "Không thấp (đơn)", True: "Thấp (đơn)"})
crosstab_rates = (
    pd.crosstab(
        [analysis_orders[column] for column in cross_index],
        analysis_orders["low_score"],
        normalize="index",
        dropna=False,
    )
    .mul(100)
    .rename(columns={False: "Không thấp (%)", True: "Thấp (%)"})
)
print("Bảng chéo: số đơn theo tình trạng giao hàng × số người bán × nhóm review")
display(crosstab_counts)
print("Bảng chéo chuẩn hóa theo hàng: tỷ lệ review thấp/không thấp (%)")
display(crosstab_rates.round(2))

# Tạo thống kê đa chỉ mục để đọc cỡ mẫu, điểm và rủi ro trong từng tổ hợp.
multi_index_summary = (
    analysis_orders.groupby(cross_index, observed=True)
    .agg(
        order_count=("order_id", "size"),
        avg_review_score=("review_score", "mean"),
        low_score_pct=("low_score", "mean"),
        avg_delay_days=("delay_days", "mean"),
    )
)
multi_index_summary["low_score_pct"] *= 100
print("Nhóm đa chỉ mục: điểm review, tỷ lệ rủi ro và quy mô mẫu")
display(multi_index_summary.round(2))

# So sánh tỷ lệ giao trễ toàn bộ với tỷ lệ trong riêng nhóm review thấp.
late_mask = analysis_orders["delivery_band"].ne("Đúng hạn/sớm")
low_mask = analysis_orders["low_score"]
late_share = late_mask.mean() * 100
late_share_among_low = late_mask[low_mask].mean() * 100
on_time_groups = multi_index_summary.xs("Đúng hạn/sớm", level="delivery_band")
single_seller_risk = on_time_groups.loc["1 người bán", "low_score_pct"]
multiple_seller_risk = on_time_groups.loc["Từ 2 người bán", "low_score_pct"]
# In các phát hiện chính; giữ lưu ý đây là liên hệ mô tả, không phải nguyên nhân.
print("KEY FINDINGS — các tỷ lệ được tính trực tiếp từ dữ liệu")
print(f"• Đơn giao trễ chiếm {late_share:.2f}% toàn bộ đơn có review, nhưng chiếm {late_share_among_low:.2f}% nhóm review thấp.")
print(
    "• Trong các đơn đúng hạn/sớm, tỷ lệ review thấp là "
    f"{multiple_seller_risk:.2f}% ở nhóm từ 2 người bán "
    f"(n={int(on_time_groups.loc['Từ 2 người bán', 'order_count']):,}) so với "
    f"{single_seller_risk:.2f}% ở nhóm 1 người bán "
    f"(n={int(on_time_groups.loc['1 người bán', 'order_count']):,})."
)
print("• Đây là các mối liên hệ có điều kiện, không chứng minh nguyên nhân; luôn đọc tỷ lệ cùng số mẫu.")
```

## Câu 6 - Phương thức thanh toán, trả góp và điểm đánh giá

Theo phương thức thanh toán đầu tiên của đơn hàng và số kỳ trả góp tối đa, so sánh số đơn, điểm trung bình, tỷ lệ đánh giá thấp và tỷ lệ đánh giá tích cực.

```python
# Tổng hợp số đơn và các chỉ số review theo phương thức thanh toán.
q06_payment = review_orders.dropna(subset=["first_payment_type"]).groupby("first_payment_type").agg(
    order_count=("order_id", "size"),
    avg_installments=("max_installments", "mean"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
    positive_pct=("positive_review", "mean"),
).sort_values("order_count", ascending=False)
# Đổi các tỷ lệ review sang phần trăm.
q06_payment[["low_score_pct", "positive_pct"]] *= 100
# Gom số kỳ trả góp thành nhóm để so sánh review giữa các mức kỳ hạn.
review_orders["installment_group"] = pd.cut(
    review_orders["max_installments"].fillna(0), bins=[-np.inf, 1, 3, 6, np.inf],
    labels=["1 kỳ", "2-3 kỳ", "4-6 kỳ", "7+ kỳ"]
)
q06_installments = review_orders.groupby("installment_group", observed=False).agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
)
# Đổi tỷ lệ review thấp sang phần trăm rồi hiển thị cả hai bảng.
q06_installments["low_score_pct"] *= 100
print("Theo phương thức thanh toán:")
display(q06_payment.round(2))
print("Theo số kỳ trả góp tối đa:")
display(q06_installments.round(2))
```

**Nhận xét phân tích và chọn bảng vẽ:** Các phương thức phổ biến có điểm gần nhau (credit card và boleto 4,16); debit card đạt 4,25 trên 1.424 đơn. Theo kỳ trả góp, điểm giảm nhẹ từ 4,19 (1 kỳ) xuống 4,08 (7+ kỳ). Chọn bảng `q06_payment` để vẽ một biểu đồ thanh ngang có ghi cỡ mẫu; bảng `q06_installments` vẫn được giữ để đối chiếu nhưng không tạo biểu đồ thứ hai.


```python
# Sắp xếp phương thức thanh toán theo điểm trung bình trước khi vẽ.
plot_data = q06_payment.sort_values("avg_review_score")
# Dùng thanh ngang để tên các phương thức dễ đọc.
fig, ax = plt.subplots(figsize=(9, 4.5))
bars = ax.barh(plot_data.index.astype(str), plot_data["avg_review_score"], color="#2878B5", label="Điểm đánh giá trung bình")
ax.bar_label(bars, labels=[f"{score:.2f} (n={count:,})" for score, count in zip(plot_data["avg_review_score"], plot_data["order_count"])], padding=4, fontsize=8)
ax.set(title="Câu 6 — Đánh giá điểm theo phương thức thanh toán", xlabel="Điểm đánh giá trung bình (1-5)", ylabel="Phương thức thanh toán", xlim=(0, 5))
ax.grid(axis="x", alpha=0.25)
ax.legend(loc="lower right")
fig.tight_layout()
# Lưu và hiển thị biểu đồ Câu 6.
save_figure("04_06_payment_vs_review_score.png")
plt.show()
plt.close(fig)
```

Chênh lệch này nhỏ hơn rõ rệt so với khác biệt giữa nhóm giao trễ và đúng hạn, còn nhóm thanh toán hiếm có ít quan sát. Vì vậy, thanh toán phù hợp làm tín hiệu bổ trợ trong phân tích và mô hình, không phải đòn bẩy vận hành chính để cải thiện hài lòng.

## Câu 7 - Xu hướng đánh giá theo tháng đặt hàng
Vì bên số liệu ở sql đã thấy nó bị thiếu dữ liệu năm 2016 nên bảng này tổng hợp số đơn, điểm trung bình và tỷ lệ đánh giá tích cực hoặc thấp theo tháng đặt hàng từ tháng 01/2017. Cỡ mẫu theo tháng giúp tránh diễn giải quá mức các giai đoạn có ít đơn.

```python
# Chỉ giữ các đơn đặt từ ngày 01/01/2017 trở đi trước khi tổng hợp theo tháng.
review_orders_from_2017 = review_orders.loc[
    review_orders["purchase_year_month"] >= pd.Timestamp("2017-01-01")
]
# Tổng hợp quy mô mẫu và chỉ số review theo từng tháng từ năm 2017.
q07 = review_orders_from_2017.groupby("purchase_year_month").agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    positive_pct=("positive_review", "mean"),
    low_score_pct=("low_score", "mean"),
).sort_index()
# Chuyển tỷ lệ tích cực/thấp thành phần trăm, làm tròn và hiển thị bảng.
q07[["positive_pct", "low_score_pct"]] *= 100
q07 = q07.round(2)
display(q07)
```

**Nhận xét phân tích và chọn bảng vẽ:** Điểm theo tháng dao động, đạt 4,32 vào 08/2017, giảm xuống 3,81 vào 03/2018 (tỷ lệ review thấp 21,09%) rồi phục hồi lên 4,21 vào 04/2018. Chọn bảng `q07` để vẽ điểm trung bình cùng số lượng đơn theo tháng, nhấn mạnh cỡ mẫu và biến động xu hướng.

```python
# Vẽ điểm trung bình theo tháng và đặt trục phụ cho số đơn.
fig, ax_score = plt.subplots(figsize=(12, 4.5))
ax_score.plot(q07.index, q07["avg_review_score"], color="#2878B5", marker="o", label="Điểm TB")
ax_score.set_ylabel("Điểm đánh giá trung bình (1-5)")
ax_score.set_ylim(0, 5)
ax_score.tick_params(axis="x", rotation=45)
# Dùng trục y thứ hai để thể hiện số đơn mà không chung thang điểm.
ax_count = ax_score.twinx()
ax_count.bar(q07.index, q07["order_count"], alpha=0.18, color="#555555", label="Số đơn")
ax_count.set_ylabel("Số đơn có đánh giá")
ax_score.set_title("Câu 7 — Điểm đánh giá theo tháng đặt hàng")
ax_score.set_xlabel("Tháng đặt hàng")
ax_score.grid(axis="y", alpha=0.25)
# Kết hợp chú giải của đường điểm và cột số đơn.
line_handles, line_labels = ax_score.get_legend_handles_labels()
bar_handles, bar_labels = ax_count.get_legend_handles_labels()
ax_score.legend(line_handles + bar_handles, line_labels + bar_labels, loc="upper left")
fig.tight_layout()
# Lưu và hiển thị biểu đồ xu hướng theo tháng.
save_figure("04_07_monthly_review_score.png")
plt.show()
plt.close(fig)
```

**Nhận xét biểu đồ:** Điểm review theo tháng dao động, đạt 4,32 vào 08/2017 rồi giảm còn 3,81 vào 03/2018, khi tỷ lệ review thấp lên 21,09%, trước khi phục hồi lên 4,21 vào tháng cuối có dữ liệu. Biến động này giúp nhận diện giai đoạn cần điều tra nhưng không xác định nguyên nhân; Olist có thể dùng giám sát theo tháng để phát hiện sớm suy giảm chất lượng và đối chiếu với thay đổi vận hành theo thời gian.

## Câu 8 - Mức hài lòng theo bang khách hàng

So sánh số đơn, điểm trung bình, tỷ lệ đánh giá tích cực hoặc thấp, tỷ lệ giao trễ và thời gian giao theo bang khách hàng.

```python
# Tổng hợp chỉ số review và giao hàng theo bang khách hàng.
q08 = review_orders.groupby("customer_state").agg(
    order_count =("order_id", "size"),
    avg_review_score = ("review_score", "mean"),
    positive_pct = ("positive_review","mean"),
    low_score_pct = ("low_score", "mean"),
    late_rate_pct = ("is_late", "mean"),
    avg_delivery_days = ("delivery_days", "mean"),
)
# Đổi các tỷ lệ sang phần trăm; xếp bang điểm thấp lên trước để rà soát.
q08[["positive_pct", "low_score_pct", "late_rate_pct"]] *= 100
q08 = q08.sort_values(["avg_review_score", "order_count"], ascending=[True, False]).round(2)
display(q08)
# Kiểm tra tương quan Spearman ở cấp bang, không diễn giải như nhân quả.
state_rho, state_p = stats.spearmanr(q08["late_rate_pct"], q08["avg_review_score"])
print(f"Spearman giữa tỷ lệ giao trễ và điểm review cấp bang: rho={state_rho:.3f}, p={format_p_value(state_p)}")
print("Tương quan trên các bang không phải kiểm định nhân quả ở cấp đơn hàng.")
```

**Nhận xét phân tích và chọn bảng vẽ:** Spearman giữa tỷ lệ giao trễ và điểm review cấp bang là -0,880 (`p < 0,001`): bang có giao trễ cao hơn thường có điểm thấp hơn. Ví dụ MA có tỷ lệ trễ 17,28%, điểm 3,82 (706 đơn), trong khi SP có tỷ lệ trễ 4,44%, điểm 4,25 (40.063 đơn). Chọn bảng `q08` để vẽ scatter: tỷ lệ trễ trên trục x, điểm trung bình trên trục y, màu biểu thị tỷ lệ review thấp và kích thước biểu thị số đơn.

```python
# Vẽ mỗi bang thành một điểm: vị trí là tỷ lệ giao trễ và điểm review.
fig, ax = plt.subplots(figsize=(9, 6))
# Kích thước biểu thị số đơn; màu biểu thị tỷ lệ review thấp.
points = ax.scatter(
    q08["late_rate_pct"], q08["avg_review_score"],
    s=np.maximum(q08["order_count"] / 100, 18), c=q08["low_score_pct"],
    cmap="coolwarm", alpha=0.8, edgecolors="white"
)
# Ghi nhãn mã bang cạnh từng điểm để xác định khu vực cần rà soát.
for state, row in q08.iterrows():
    ax.annotate(state, (row["late_rate_pct"], row["avg_review_score"]),
                fontsize=8, xytext=(3, 3), textcoords="offset points")
ax.set(title="Câu 8 — Tỷ lệ giao trễ và điểm đánh giá theo bang",
       xlabel="Tỷ lệ đơn giao trễ (%)", ylabel="Điểm trung bình (1-5)", ylim=(0, 5))
ax.grid(alpha=0.25)
# Thêm thang màu cho review thấp và chú giải kích thước cho số đơn.
fig.colorbar(points, ax=ax, label="Tỷ lệ đánh giá thấp (≤2 sao, %)")
size_handles, size_labels = points.legend_elements(
    prop="sizes", num=3, func=lambda value: value * 100, fmt="{x:.0f}"
)
ax.legend(size_handles, size_labels, title="Số đơn", loc="lower left")
fig.tight_layout()
# Lưu và hiển thị biểu đồ so sánh giữa các bang.
save_figure("04_08_review_by_state.png")
plt.show()
plt.close(fig)
```

**Nhận xét:** Ở cấp bang, tỷ lệ giao trễ và điểm review có tương quan Spearman âm mạnh (p = −0,880; p < 0,001), cho thấy các bang trễ nhiều thường có điểm thấp hơn. Đây là quan hệ tổng hợp theo bang, không thể suy rộng trực tiếp thành quan hệ ở cấp đơn hàng hoặc xem là bằng chứng nhân quả. Kết quả giúp Olist khoanh vùng cần rà soát tuyến và đối tác logistics, đồng thời phải ưu tiên bang có đủ số đơn để tránh bị chi phối bởi mẫu nhỏ.
