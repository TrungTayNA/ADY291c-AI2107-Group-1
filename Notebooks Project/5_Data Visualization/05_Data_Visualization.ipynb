# Trực quan hóa phân tích trải nghiệm khách hàng

Nguồn dữ liệu dùng chung là SQLite view `v_review_order_features`. Matplotlib/Seaborn được dùng cho biểu đồ tĩnh và Plotly cho biểu đồ tương tác có hover, zoom và pan.

**Mục tiêu:** Trực quan hóa và diễn giải kết quả phân tích ở Phần 4 theo cùng tám câu hỏi, đơn vị phân tích và nguồn dữ liệu, làm rõ phân bố review score và các mối liên hệ với trải nghiệm giao hàng, phí vận chuyển, độ phức tạp đơn hàng, thanh toán, thời điểm và khu vực. Các biểu đồ hỗ trợ nhận diện yếu tố vận hành liên quan đến review thấp và định hướng phát hiện sớm đơn hàng có nguy cơ như mục tiêu dự án ở Chương 1; các mối liên hệ không được diễn giải là quan hệ nhân quả.


```python
from pathlib import Path
import sqlite3

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import plotly.express as px
import seaborn as sns

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams.update({"figure.dpi": 130, "axes.titleweight": "bold"})


def find_project_root(start_dir: Path) -> Path:
    for candidate in (start_dir, *start_dir.parents):
        if (
            (candidate / "data" / "olist.db").is_file()
            and (candidate / "Notebooks").is_dir()
            and (candidate / "sql" / "03_queries.sql").is_file()
        ):
            return candidate
    raise FileNotFoundError(
        f"Không tìm thấy thư mục gốc từ {start_dir}. "
        "Hãy chạy scripts/database/build_db.py trước."
    )


BASE_DIR = find_project_root(Path.cwd().resolve())
DB_PATH = BASE_DIR / "data" / "olist.db"
FIG_DIR = BASE_DIR / "reports" / "figures"
FIG_DIR.mkdir(parents=True, exist_ok=True)

with sqlite3.connect(DB_PATH) as conn:
    available_views = {
        row[0]
        for row in conn.execute("SELECT name FROM sqlite_master WHERE type='view'")
    }
    if "v_review_order_features" not in available_views:
        raise RuntimeError(
            "Thiếu view v_review_order_features. "
            "Hãy chạy scripts/database/build_db.py để dựng lại database."
        )
    reviews = pd.read_sql_query(
        "SELECT * FROM v_review_order_features ORDER BY purchase_year_month, order_id",
        conn,
    )

reviews["is_late"] = reviews["is_late"].astype(bool)
reviews["low_score"] = reviews["review_score"] <= 2
reviews["positive_review"] = reviews["review_score"] >= 4
reviews["purchase_year_month"] = pd.to_datetime(
    reviews["purchase_year_month"], format="%Y-%m"
)

def save_figure(fig, filename):
    path = FIG_DIR / filename
    fig.savefig(path, bbox_inches="tight", dpi=180)
    plt.show()
    plt.close(fig)

print(f"Database: {DB_PATH}")
print(f"Đơn hàng có review: {len(reviews):,}")

```

## Câu 1 — Phân bố và mất cân bằng review score

Trực quan hóa kết quả của Câu 1 ở Phần 4: số lượng và tỷ trọng đơn ở từng mức 1–5 sao. Tỷ trọng giúp nhìn rõ mức độ mất cân bằng của nhãn review.


```python
q03 = reviews.groupby("review_score").size().reindex(range(1, 6), fill_value=0)
fig, ax = plt.subplots(figsize=(8, 4.5))
bars = ax.bar(q03.index.astype(str), q03.values, color="#2878B5")
ax.bar_label(bars, labels=[f"{v / q03.sum():.1%}" for v in q03.values], padding=3)
ax.set(title="Câu 1 — Phân bố review score", xlabel="Điểm review (sao)", ylabel="Số đơn có review (đơn)")
ax.margins(y=0.15)
fig.tight_layout()
save_figure(fig, "05_01_review_score_distribution.png")
```

**Nhận xét:** Điểm trung bình chung là **4,16/5**; 5 sao chiếm **59,25%** (56.480 đơn), trong khi review 1–2 sao chiếm **12,77%**. Như vậy, mức trung bình cao vẫn che khuất một nhóm khách hàng không hài lòng đáng kể. Với doanh nghiệp, theo dõi riêng nhóm điểm thấp giúp tránh bỏ sót trải nghiệm xấu khi chỉ nhìn vào điểm tổng quát. Đây là cơ sở mô tả ban đầu cho mục tiêu Chương 1: nhận diện sớm đơn hàng có nguy cơ review thấp.


## Câu 2 — Giao trễ và giao đúng hạn liên quan đến review thế nào?

Trực quan hóa kết quả của Câu 2 ở Phần 4: so sánh điểm trung bình, tỷ lệ review tích cực (≥4 sao) và review thấp (≤2 sao) giữa hai nhóm giao hàng.


```python
q04 = reviews.groupby("is_late").agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    positive_pct=("positive_review", "mean"),
    low_score_pct=("low_score", "mean"),
).rename(index={False: "Đúng hạn hoặc sớm", True: "Giao trễ"})
q04[["positive_pct", "low_score_pct"]] *= 100
print(q04.round(2))

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
q04["avg_review_score"].plot(kind="bar", ax=axes[0], color=["#4C9F70", "#D9534F"])
axes[0].set(title="Điểm review trung bình", ylabel="Điểm (sao, 1–5)", xlabel="Tình trạng giao hàng")
axes[0].set_ylim(0, 5)
axes[0].tick_params(axis="x", rotation=15)
q04[["positive_pct", "low_score_pct"]].plot(kind="bar", ax=axes[1], color=["#4C9F70", "#D9534F"])
axes[1].set(title="Tỷ lệ review theo nhóm", ylabel="Tỷ lệ review (%)", xlabel="Tình trạng giao hàng")
axes[1].tick_params(axis="x", rotation=15)
axes[1].legend(["Tích cực (≥4)", "Thấp (≤2)"])
fig.suptitle("Câu 2 — Tình trạng giao hàng và review score")
fig.tight_layout()
save_figure(fig, "05_02_late_vs_review_score.png")
```

**Nhận xét:** Nhóm giao trễ có điểm trung bình **2,27**, thấp hơn nhóm đúng hạn/sớm (**4,29**); tỷ lệ review thấp là **62,37%** so với **9,23%**. Ngược lại, tỷ lệ review tích cực ở nhóm đúng hạn là **82,71%**, cao hơn nhiều so với **26,72%** ở nhóm trễ. Chênh lệch này cho thấy tình trạng giao hàng là tín hiệu vận hành quan trọng để ưu tiên theo dõi và can thiệp. Kết quả hỗ trợ trực tiếp bài toán Chương 1 về xác định đơn hàng có nguy cơ không hài lòng, nhưng chỉ thể hiện mối liên hệ chứ không chứng minh giao trễ là nguyên nhân duy nhất.


## Câu 3 — Mức độ giao trễ và review score

Trực quan hóa kết quả của Câu 3 ở Phần 4, dùng các nhóm trễ: sớm/đúng hạn, trễ 1–3 ngày, 4–7 ngày và trên 7 ngày.


```python
delay_labels = ["Sớm hoặc đúng hạn", "Trễ 1–3 ngày", "Trễ 4–7 ngày", "Trễ trên 7 ngày"]
reviews["delay_group"] = pd.cut(
    reviews["delay_days"], [-np.inf, 0, 3, 7, np.inf], labels=delay_labels
)
q05 = reviews.groupby("delay_group", observed=False).agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    positive_pct=("positive_review", "mean"),
)
q05["positive_pct"] *= 100
print(q05.round(2))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(q05.index.astype(str), q05["avg_review_score"], marker="o", linewidth=2.5, color="#D9534F", label="Điểm review trung bình")
ax.set(title="Câu 3 — Review score theo số ngày trễ", xlabel="Mức trễ so với ngày dự kiến", ylabel="Điểm review trung bình (sao, 1–5)")
ax.set_ylim(0, 5)
ax.grid(axis="y", alpha=0.3)
ax2 = ax.twinx()
ax2.bar(q05.index.astype(str), q05["order_count"], alpha=0.2, color="#2878B5", label="Số đơn")
ax2.set_ylabel("Số đơn có review (đơn)")
handles, labels = ax.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax.legend(handles + handles2, labels + labels2, loc="best")
fig.tight_layout()
save_figure(fig, "05_03_delay_vs_review_score.png")
```

**Nhận xét:** Điểm trung bình giảm theo mức trễ: **3,29** ở nhóm trễ 1–3 ngày, **2,11** ở nhóm 4–7 ngày và **1,69** ở nhóm trễ trên 7 ngày. Số đơn trễ trên 7 ngày là **2.769**, trong đó chỉ **11,70%** review tích cực. Mức giảm theo thời gian trễ cho thấy các đơn giao chậm kéo dài cần được ưu tiên giám sát hơn các đơn chỉ chậm nhẹ. Đây là tín hiệu có thể hỗ trợ cảnh báo sớm và cải thiện logistics như mục tiêu Chương 1, dù biểu đồ không khẳng định quan hệ nhân quả.


#### Phân phối độ lệch giao hàng theo review score

Boxplot bổ sung góc nhìn về trung vị, khoảng tứ phân vị và độ phân tán của số ngày giao so với ngày dự kiến trong từng mức review.

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=reviews, x="review_score", y="delay_days", color="#78A6C8", showfliers=False, ax=ax)
ax.axhline(0, color="#D9534F", linestyle="--", linewidth=1.5, label="Đúng ngày dự kiến")
ax.set(
    title="Phân phối số ngày giao so với hạn dự kiến theo review score",
    xlabel="Điểm review (sao, 1–5)",
    ylabel="Số ngày so với ngày dự kiến (ngày; âm = giao sớm)",
)
ax.legend()
fig.tight_layout()
save_figure(fig, "05_09_delay_boxplot_by_review_score.png")
```

**Nhận xét:** Trung vị số ngày so với hạn dự kiến là **−7 ngày** ở nhóm 1 sao và **−13 ngày** ở nhóm 5 sao; nhóm 1 sao có phân phối dịch về phía giao muộn hơn. Khoảng tứ phân vị của nhóm 1 sao kéo dài đến **+6 ngày**, trong khi 75% nhóm 5 sao được giao sớm ít nhất **8 ngày** theo mốc dự kiến. Chênh lệch về phân phối cho thấy việc chỉ so sánh trung bình có thể bỏ qua mức độ biến thiên trong trải nghiệm giao hàng. Đây là một tín hiệu hữu ích để Olist ưu tiên giám sát đơn có nguy cơ giao sát hạn hoặc trễ, phù hợp với mục tiêu cảnh báo sớm ở Chương 1.


## Câu 4 — Phí vận chuyển, tỷ lệ phí/giá và review score

Trực quan hóa kết quả của Câu 4 ở Phần 4: chia đơn thành bốn quartile theo tổng phí ship và theo tỷ lệ phí ship/giá trị hàng, so sánh điểm review trung bình và hiển thị cỡ mẫu.


```python
quartiles = ["Q1 thấp", "Q2", "Q3", "Q4 cao"]
reviews["freight_quartile"] = pd.qcut(reviews["total_freight"].rank(method="first"), 4, labels=quartiles)
valid_ratio = reviews.dropna(subset=["freight_to_price_pct"]).copy()
valid_ratio["ratio_quartile"] = pd.qcut(
    valid_ratio["freight_to_price_pct"].rank(method="first"), 4, labels=quartiles
)
q06_freight = reviews.groupby("freight_quartile", observed=False).agg(
    order_count=("order_id", "size"), avg_review_score=("review_score", "mean")
)
q06_ratio = valid_ratio.groupby("ratio_quartile", observed=False).agg(
    order_count=("order_id", "size"), avg_review_score=("review_score", "mean")
)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5), sharey=True)
for ax, summary, title in [
    (axes[0], q06_freight, "Theo tổng phí ship"),
    (axes[1], q06_ratio, "Theo tỷ lệ phí/giá"),
]:
    bars = ax.bar(summary.index.astype(str), summary["avg_review_score"], color="#8172B2")
    ax.bar_label(bars, labels=[f"n={n:,}" for n in summary["order_count"]], padding=3, fontsize=8)
    ax.set(title=title, xlabel="Nhóm tứ phân vị phí vận chuyển", ylim=(0, 5))
    ax.tick_params(axis="x", rotation=15)
    ax.grid(axis="y", alpha=0.25)
axes[0].set_ylabel("Điểm review trung bình (sao, 1–5)")
fig.suptitle("Câu 4 — Phí vận chuyển và review score")
fig.tight_layout()
save_figure(fig, "05_04_freight_vs_review_score.png")
```

**Nhận xét:** Điểm trung bình giảm từ **4,32** ở quartile phí ship thấp xuống **3,94** ở quartile cao; trong khi theo tỷ lệ phí/giá mức chênh nhỏ hơn (**4,20** so với **4,11**). Mỗi nhóm có khoảng **23.832–23.833 đơn**, nên các mức chênh lệch phản ánh những nhóm có quy mô tương đương. Phí vận chuyển có thể là một phần của trải nghiệm cần xem xét, nhưng mức chênh nhỏ ở tỷ lệ phí/giá cho thấy không nên dùng riêng yếu tố này để giải thích review thấp. Kết quả gợi ý Olist nên kết hợp chi phí giao hàng với các tín hiệu vận hành khác khi tìm đơn có rủi ro, không diễn giải đây là quan hệ nhân quả.


#### Biểu đồ scatter tương tác: giá trị giỏ hàng và phí vận chuyển

Biểu đồ Plotly lấy mẫu tối đa 5.000 đơn để dễ thao tác; rê chuột xem thông tin đơn, đồng thời có thể phóng to, thu nhỏ và kéo vùng biểu đồ.

```python
scatter_source = reviews.dropna(subset=["item_total_sum", "total_freight", "review_score"])
scatter_source = scatter_source.loc[scatter_source["item_total_sum"] > 0]
scatter_data = scatter_source.sample(n=min(5000, len(scatter_source)), random_state=42)
interactive_scatter = px.scatter(
    scatter_data,
    x="item_total_sum",
    y="total_freight",
    color="review_score",
    color_continuous_scale="RdYlGn",
    range_color=(1, 5),
    hover_data={"order_id": True, "customer_state": True, "item_count": True},
    labels={
        "item_total_sum": "Tổng giá trị hàng (R$)",
        "total_freight": "Tổng phí vận chuyển (R$)",
        "review_score": "Điểm review (sao, 1–5)",
    },
    title=f"Giá trị giỏ hàng và phí vận chuyển theo review (n={len(scatter_data):,})",
    render_mode="webgl",
)
interactive_scatter.update_traces(marker={"size": 6, "opacity": 0.65})
interactive_scatter.update_layout(coloraxis_colorbar={"title": "Điểm review<br>(sao)"})
interactive_scatter.show()
```

**Nhận xét:** Scatter lấy mẫu **5.000 đơn** cho thấy phí vận chuyển thường tăng cùng giá trị hàng trong giỏ; trên toàn bộ dữ liệu, tương quan giữa hai biến là khoảng **0,49**. Tương quan giữa phí vận chuyển và điểm review chỉ khoảng **−0,09**, nên mức phí riêng lẻ không giải thích tốt review thấp. Màu sắc và tooltip giúp kiểm tra các điểm bất thường theo review, bang và số mặt hàng mà không che khuất đặc điểm từng đơn. Điều này gợi ý doanh nghiệp nên kết hợp chi phí với thời gian giao và độ phức tạp đơn hàng khi xây dựng cảnh báo rủi ro theo Chương 1.


## Câu 5 — Độ phức tạp đơn hàng và review score

Trực quan hóa kết quả của Câu 5 ở Phần 4: so sánh điểm review và tỷ lệ review thấp theo số mặt hàng trong giỏ và số người bán trong đơn.


```python
reviews["basket_group"] = np.select(
    [reviews["item_count"].eq(1), reviews["item_count"].between(2, 3)],
    ["1 mặt hàng", "2–3 mặt hàng"], default="4+ mặt hàng"
)
reviews["seller_group"] = np.where(reviews["seller_count"].eq(1), "1 người bán", "Nhiều người bán")
q07 = reviews.groupby(["basket_group", "seller_group"]).agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
)
q07["low_score_pct"] *= 100
print(q07.round(2))

score = q07["avg_review_score"].unstack("seller_group")
fig, ax = plt.subplots(figsize=(9, 4.5))
score.plot(kind="bar", ax=ax, color=["#2878B5", "#E07B39"])
ax.set(title="Câu 5 — Review theo quy mô giỏ và số người bán", xlabel="Quy mô giỏ hàng (mặt hàng)", ylabel="Điểm review trung bình (sao, 1–5)")
ax.set_ylim(0, 5)
ax.tick_params(axis="x", rotation=0)
ax.legend(title="Người bán")
ax.grid(axis="y", alpha=0.25)
fig.tight_layout()
save_figure(fig, "05_05_order_complexity_vs_review_score.png")
```

**Nhận xét:** Điểm giảm khi giỏ hàng lớn hơn và có nhiều người bán: đơn một mặt hàng/một người bán đạt **4,21**, còn nhóm 4+ mặt hàng/nhiều người bán đạt **2,49**; tỷ lệ review thấp ở nhóm sau là **59,65%**. Tuy nhiên, nhóm này chỉ có **114 đơn**, nên kết quả cần được diễn giải thận trọng. Độ phức tạp cao có thể khiến việc phối hợp người bán và giao hàng khó kiểm soát hơn, vì vậy nhóm này đáng được theo dõi thêm. Điều đó phù hợp với vấn đề Chương 1 về trải nghiệm đơn hàng phụ thuộc vào nhiều người bán và các bên logistics.


## Câu 6 — Phương thức thanh toán, trả góp và review score

Trực quan hóa kết quả của Câu 6 ở Phần 4: đối chiếu điểm trung bình, tỷ lệ review thấp và cỡ mẫu theo `first_payment_type`.


```python
q08 = reviews.dropna(subset=["first_payment_type"]).groupby("first_payment_type").agg(
    order_count=("order_id", "size"),
    avg_installments=("max_installments", "mean"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
).sort_values("order_count", ascending=False)
q08["low_score_pct"] *= 100
print(q08.round(2))

fig, ax = plt.subplots(figsize=(9, 4.5))
plot_data = q08.sort_values("avg_review_score")
bars = ax.barh(plot_data.index, plot_data["avg_review_score"], color="#2878B5")
ax.bar_label(bars, labels=[f"{score:.2f} (n={n:,})" for score, n in zip(plot_data["avg_review_score"], plot_data["order_count"])], padding=4)
ax.set(title="Câu 6 — Review score theo phương thức thanh toán", xlabel="Điểm review trung bình (sao, 1–5)", xlim=(0, 5))
ax.set_ylabel("Phương thức thanh toán")
ax.grid(axis="x", alpha=0.25)
fig.tight_layout()
save_figure(fig, "05_06_payment_vs_review_score.png")
```

**Nhận xét:** Các phương thức phổ biến có điểm trung bình khá gần nhau: credit card và boleto đều **4,16**, voucher **4,11**; debit card đạt **4,25** trên **1.424 đơn**. Nhóm debit card có tỷ lệ review thấp **10,88%**, thấp hơn credit card (**12,84%**), nhưng khác biệt quan sát được không đủ để kết luận phương thức thanh toán làm thay đổi mức hài lòng. Vì vậy, thanh toán có vẻ là tín hiệu yếu hơn giao hàng trong các so sánh mô tả ở notebook này. Khi xây dựng cảnh báo rủi ro theo Chương 1, phương thức thanh toán nên được kết hợp với đặc trưng vận hành thay vì dùng làm căn cứ can thiệp riêng.


## Câu 7 — Xu hướng review theo tháng đặt hàng

Trực quan hóa kết quả của Câu 7 ở Phần 4: thể hiện điểm review trung bình, tỷ lệ review thấp và số đơn theo tháng đặt hàng. Tháng đặt hàng là mốc nhất quán trong view; không diễn giải nó thành tháng review được gửi.


```python
q09 = reviews.groupby("purchase_year_month").agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
).sort_index()
q09["low_score_pct"] *= 100

fig, (ax_score, ax_count) = plt.subplots(
    2, 1, figsize=(12, 7), sharex=True, gridspec_kw={"height_ratios": [2, 1]}
)
ax_score.plot(q09.index, q09["avg_review_score"], marker="o", color="#2878B5", label="Điểm review trung bình")
ax_score.set(title="Câu 7 — Review score theo tháng đặt hàng", ylabel="Điểm review trung bình (sao, 1–5)", ylim=(0, 5))
ax_score.grid(axis="y", alpha=0.25)
ax_rate = ax_score.twinx()
ax_rate.plot(q09.index, q09["low_score_pct"], marker="s", color="#D9534F", label="Review thấp (≤2 sao)")
ax_rate.set_ylabel("Review thấp (≤2 sao, %)")
ax_rate.set_ylim(0, 100)
lines = ax_score.get_lines() + ax_rate.get_lines()
ax_score.legend(lines, [line.get_label() for line in lines], loc="best")
ax_count.bar(q09.index, q09["order_count"], width=20, color="#8172B2", alpha=0.8)
ax_count.set_ylabel("Số đơn có review (đơn)")
ax_count.set_xlabel("Tháng đặt hàng")
ax_count.grid(axis="y", alpha=0.25)
ax_count.tick_params(axis="x", rotation=45)
fig.tight_layout()
save_figure(fig, "05_07_monthly_review_score.png")
```

**Nhận xét:** Điểm theo tháng dao động thay vì tăng đều: từ **4,32** vào 08/2017 giảm xuống **3,81** vào 03/2018, rồi phục hồi lên **4,31** vào 06/2018. Tỷ lệ review thấp cũng tăng lên **21,09%** vào 03/2018 trước khi giảm về khoảng **9,5–10,0%** vào 06–08/2018. Biến động theo thời gian có thể giúp quản lý nhận ra giai đoạn cần điều tra vận hành sâu hơn, nhưng các tháng đầu có ít quan sát nên cần đọc cùng cỡ mẫu. Theo Chương 1, việc theo dõi xu hướng này hỗ trợ giám sát chất lượng dịch vụ và định hướng cảnh báo, không chứng minh tháng đặt hàng là nguyên nhân của điểm thấp.


## Câu 8 — Mức hài lòng theo bang khách hàng

Trực quan hóa kết quả của Câu 8 ở Phần 4: so sánh điểm review, tỷ lệ giao trễ và tỷ lệ review thấp theo bang, chỉ giữ bang có ít nhất 100 đơn như điều kiện lọc đã dùng ở Phần 4. Biểu đồ thanh ngang thể hiện mức độ khác biệt giữa các bang; cần cân nhắc cỡ mẫu khi diễn giải và không xem liên hệ địa lý là quan hệ nhân quả.


```python
q10 = reviews.groupby("customer_state").agg(
    order_count=("order_id", "size"),
    avg_review_score=("review_score", "mean"),
    low_score_pct=("low_score", "mean"),
    late_rate_pct=("is_late", "mean"),
)
q10 = q10.loc[q10["order_count"] >= 100].sort_values("avg_review_score")
q10[["low_score_pct", "late_rate_pct"]] *= 100
print(q10.round(2))

fig, axes = plt.subplots(1, 2, figsize=(12, 9), sharey=True)
colors = plt.cm.coolwarm(plt.Normalize(q10["low_score_pct"].min(), q10["low_score_pct"].max())(q10["low_score_pct"]))
axes[0].barh(q10.index, q10["avg_review_score"], color=colors)
axes[0].set(title="Điểm review (màu = tỷ lệ thấp)", xlabel="Điểm review trung bình (sao, 1–5)", xlim=(0, 5))
axes[0].set_ylabel("Bang khách hàng")
axes[0].grid(axis="x", alpha=0.25)
axes[1].barh(q10.index, q10["late_rate_pct"], color="#8172B2")
axes[1].set(title="Tình trạng giao hàng", xlabel="Tỷ lệ đơn giao trễ (%)")
axes[1].grid(axis="x", alpha=0.25)
colorbar = fig.colorbar(plt.cm.ScalarMappable(norm=plt.Normalize(q10["low_score_pct"].min(), q10["low_score_pct"].max()), cmap="coolwarm"), ax=axes[0], pad=0.02)
colorbar.set_label("Review thấp (≤2 sao, %)")
fig.suptitle("Câu 8 — Mức hài lòng và giao trễ theo bang", y=1.01)
fig.tight_layout()
save_figure(fig, "05_08_review_by_state.png")
```

**Nhận xét:** Trong các bang có ít nhất 100 đơn, MA có điểm thấp nhất (**3,82**, 706 đơn) và tỷ lệ giao trễ **17,28%**; SP có điểm cao nhất (**4,25**, 40.063 đơn) với tỷ lệ trễ **4,44%**. Mẫu của các bang rất khác nhau, nên cần kết hợp cỡ mẫu khi so sánh. Khác biệt địa lý giúp khoanh vùng nơi cần rà soát tuyến giao hàng, nhưng không đủ để kết luận vị trí là nguyên nhân của đánh giá thấp. Đây là thông tin để định hướng phân tích rủi ro theo khu vực như đề cập ở Chương 1, không phải bằng chứng nhân quả.
